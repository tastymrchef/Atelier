```python
from src.curator.workflow import curator_workflow_graph,selection_node,get_metadata

state = {"user_query": "rainy nights with moon", "candidates": res["candidates"]}
output = selection_node(state)
draft = output["selection_draft"]

print(draft.exhibit_title)
print(draft.selected_object_ids)
print(draft.narrative)
```

```python
import pandas as pd
for oid in draft.selected_object_ids:
    row = get_metadata(oid)
    title = row["Title"] if pd.notna(row["Title"]) else row["Object Name"]
    english_part = title.split("|")[-1] if isinstance(title, str) and "|" in title else title
    mentioned = isinstance(english_part, str) and english_part.split(",")[0].lower() in draft.narrative.lower()
    print(oid, "->", title, "| mentioned:", mentioned)
```
